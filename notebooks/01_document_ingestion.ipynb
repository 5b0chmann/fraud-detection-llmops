```python
import pandas as pd
from pathlib import Path
from pypdf import PdfReader

# Configure Data Directory
RAW_DATA_DIR = Path("data/raw")
RAW_DATA_DIR.mkdir(parents=True, exist_ok=True)
print("Raw data directory:")
print(RAW_DATA_DIR.resolve())

# PDF Text Extraction Function
def extract_pdf_text(pdf_path):
  reader = PdfReader(pdf_path)
  text = ""
  for page in reader.pages:
    page_text = page.extract_text()
    if page_text:
      text += page_text + "\n"
  return text

# Discover PDF Files
pdf_files = list(RAW_DATA_DIR.glob("*.pdf"))
print(f"Found {len(pdf_files)} PDF files")
pdf_files

# Extract Document Content
documents = []
for pdf_file in pdf_files:
  text = extract_pdf_text(pdf_file)
  documents.append({"file_name": pdf_file.name, "text": text})
documents_df = pd.DataFrame(documents)
documents_df.head()

# Analyze Document Length
documents_df["characters"] = (documents_df["text"].str.len())
documents_df[["file_name", "characters"]]
```

## Conclusion

Documents were successfully loaded and converted into plain text.
